```python
import ee
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats
```

```python
# Initialize Earth Engine

try:
    ee.Initialize()
    print(" Earth Engine initialized successfully\n")
except:
    print("Authenticating Earth Engine...")
    ee.Authenticate()
    ee.Initialize()
    print(" Earth Engine initialized successfully\n")
```

```python
#  Lake Selection - I chose Lake Tahoe in  Nevada

# Lake Tahoe center coordinates
lon = -120.0324
lat = 39.0968


# STEP 2: Create Region of Interest (ROI)

print("Creating Region of Interest (ROI)...")
point = ee.Geometry.Point([lon, lat])
buffered = point.buffer(200)  # 200m radius
roi = buffered.bounds()

print(f" ROI created: 200m buffer (400m diameter)")
print(f"  Center: {lat}°N, {lon}°W\n")
```

```python

#Cloud/Snow Masking Functions
def mask_landsat457_sr(image):
    """Cloud and snow masking for Landsat 4, 5, 7"""
    qa = image.select('QA_PIXEL')
    
    # Bit masks
    cloud_bit = 1 << 3
    cloud_shadow_bit = 1 << 4
    snow_bit = 1 << 5
    cirrus_bit = 1 << 2
    
    # Cloud confidence (bits 8-9)
    cloud_confidence = qa.rightShift(8).bitwiseAnd(3)
    
    # Create mask
    mask = qa.bitwiseAnd(cloud_bit).eq(0) \
        .And(qa.bitwiseAnd(cloud_shadow_bit).eq(0)) \
        .And(qa.bitwiseAnd(snow_bit).eq(0)) \
        .And(qa.bitwiseAnd(cirrus_bit).eq(0)) \
        .And(cloud_confidence.lt(2))
    
    return image.updateMask(mask)

def mask_landsat89_sr(image):
    """Cloud and snow masking for Landsat 8, 9"""
    qa = image.select('QA_PIXEL')
    
    # Bit masks
    cloud_bit = 1 << 3
    cloud_shadow_bit = 1 << 4
    snow_bit = 1 << 5
    cirrus_bit = 1 << 2
    
    # Cloud confidence
    cloud_confidence = qa.rightShift(8).bitwiseAnd(3)
    
    # Create mask
    mask = qa.bitwiseAnd(cloud_bit).eq(0) \
        .And(qa.bitwiseAnd(cloud_shadow_bit).eq(0)) \
        .And(qa.bitwiseAnd(snow_bit).eq(0)) \
        .And(qa.bitwiseAnd(cirrus_bit).eq(0)) \
        .And(cloud_confidence.lt(2))
    
    return image.updateMask(mask)

def apply_scale_offset_L457(image):
    """Apply scale and offset for Landsat 4, 5, 7"""
    thermal = image.select('ST_B6') \
        .multiply(0.00341802) \
        .add(149.0) \
        .rename('temperature_K')
    return image.addBands(thermal)

def apply_scale_offset_L89(image):
    """Apply scale and offset for Landsat 8, 9"""
    thermal = image.select('ST_B10') \
        .multiply(0.00341802) \
        .add(149.0) \
        .rename('temperature_K')
    return image.addBands(thermal)

```

```python
# Temperature Calculation 
def add_temperature_stats(image):
    """
    Calculate mean temperature over ROI and add as image properties
    Handles null values properly
    """
    # Get the masked temperature band
    temp_band = image.select('temperature_K')
    
    # Calculate statistics
    stats = temp_band.reduceRegion(
        reducer=ee.Reducer.mean().combine(
            reducer2=ee.Reducer.count(),
            sharedInputs=True
        ),
        geometry=roi,
        scale=30,
        maxPixels=1e9
    )
    
    mean_temp = stats.get('temperature_K_mean')
    pixel_count = stats.get('temperature_K_count')
    
    # Calculate Celsius only if mean_temp is not null
    mean_temp_c = ee.Algorithms.If(
        ee.Algorithms.IsEqual(mean_temp, None),
        None,
        ee.Number(mean_temp).subtract(273.15)
    )
    
    # Add properties to image (server-side)
    return image.set({
        'mean_temp_K': mean_temp,
        'mean_temp_C': mean_temp_c,
        'valid_pixels': pixel_count,
        'date_string': image.date().format('YYYY-MM-dd'),
        'timestamp': image.date().millis()
    })
```

```python
#Process Each Landsat Mission


print("Processing Landsat Collections...")
print("-" * 70)

all_temperature_data = []

# Define satellite collections
satellites = {
    'Landsat 4': {
        'collection': 'LANDSAT/LT04/C02/T1_L2',
        'mask_func': mask_landsat457_sr,
        'scale_func': apply_scale_offset_L457,
        'years': '1982-1993'
    },
    'Landsat 5': {
        'collection': 'LANDSAT/LT05/C02/T1_L2',
        'mask_func': mask_landsat457_sr,
        'scale_func': apply_scale_offset_L457,
        'years': '1984-2012'
    },
    'Landsat 7': {
        'collection': 'LANDSAT/LE07/C02/T1_L2',
        'mask_func': mask_landsat457_sr,
        'scale_func': apply_scale_offset_L457,
        'years': '1999-present'
    },
    'Landsat 8': {
        'collection': 'LANDSAT/LC08/C02/T1_L2',
        'mask_func': mask_landsat89_sr,
        'scale_func': apply_scale_offset_L89,
        'years': '2013-present'
    },
    'Landsat 9': {
        'collection': 'LANDSAT/LC09/C02/T1_L2',
        'mask_func': mask_landsat89_sr,
        'scale_func': apply_scale_offset_L89,
        'years': '2021-present'
    }
}

# Process each satellite
for sat_name, sat_info in satellites.items():
    print(f"\n{sat_name} ({sat_info['years']}):")
    print(f"  Loading collection...")
    
    # Load collection
    collection = ee.ImageCollection(sat_info['collection']) \
        .filterBounds(roi)
    
    total_images = collection.size().getInfo()
    print(f"  Total images available: {total_images}")
    
    if total_images == 0:
        print(f"   No images found")
        continue
    
    # Apply cloud/snow masking and temperature conversion (SERVER-SIDE)
    print(f"  Applying masks and calculating temperatures...")
    processed = collection \
        .map(sat_info['mask_func']) \
        .map(sat_info['scale_func']) \
        .map(add_temperature_stats)
    
    # Filter out completely cloudy images (where mean_temp_K is null)
    valid_images = processed.filter(ee.Filter.notNull(['mean_temp_K']))
    
    print(f"  Counting valid images...")
    valid_count = valid_images.size().getInfo()
    cloudy_count = total_images - valid_count
    
    print(f"   Valid images: {valid_count}")
    print(f"   Cloudy/masked images: {cloudy_count}")
    
    # Extract properties to list
    if valid_count > 0:
        print(f"  Downloading results...")
        
        def extract_properties(image):
            return ee.Feature(None, {
                'satellite': sat_name,
                'date': image.get('date_string'),
                'timestamp': image.get('timestamp'),
                'mean_temp_K': image.get('mean_temp_K'),
                'mean_temp_C': image.get('mean_temp_C'),
                'valid_pixels': image.get('valid_pixels')
            })
        
        features = valid_images.map(extract_properties)
        feature_collection = ee.FeatureCollection(features)
        
        
        try:
            results = feature_collection.getInfo()
            
            # Extract data from results
            for feature in results['features']:
                props = feature['properties']
                if props['mean_temp_K'] is not None:
                    all_temperature_data.append(props)
            
            print(f"   Downloaded {len(results['features'])} measurements")
        except Exception as e:
            print(f"   Error downloading: {str(e)}")
            continue

print("\n" + "=" * 70)
print(f"TOTAL VALID MEASUREMENTS: {len(all_temperature_data)}")
print("=" * 70 + "\n")


# Export to CSV
csv_filename = 'lake_tahoe_temperature_data.csv'
print(f"Exporting data to {csv_filename}...")

# Create DataFrame
df = pd.DataFrame(all_temperature_data)
df['date'] = pd.to_datetime(df['date'])
df = df.sort_values('date')

# Save to CSV
df.to_csv(csv_filename, index=False)
print(f" Data exported successfully\n")

# Display sample data
print("Sample of collected data:")
print(df.head(10).to_string(index=False))
print(f"\n... {len(df)} total records ...\n")
```

```python
# STEP 7: Statistical Analysis
print("=" * 70)
print("STATISTICAL ANALYSIS")
print("=" * 70)

print("\nTemperature Statistics:")
print(df[['mean_temp_K', 'mean_temp_C']].describe())

print("\nData by Satellite:")
print(df.groupby('satellite').agg({
    'mean_temp_C': ['count', 'mean', 'std', 'min', 'max']
}).round(2))

# Calculate annual means
df['year'] = df['date'].dt.year
annual_means = df.groupby('year')['mean_temp_C'].mean()

# Calculate trend
years = annual_means.index.values
temps = annual_means.values

#Calculating p-value to test significance
slope, intercept, r_value, p_value, std_err = stats.linregress(years, temps)

print(f"\n" + "=" * 70)
print("TREND ANALYSIS")
print("=" * 70)
print(f"Temperature trend: {slope:.4f} °C/year")
print(f"Total change ({years[0]}-{years[-1]}): {slope * (years[-1]-years[0]):.2f} °C")
print(f"P-value: {p_value:.6f}")

if p_value < 0.05:
    print(f"\n STATISTICALLY SIGNIFICANT warming trend detected!")
else:
    print(f"\n Trend is not statistically significant (p > 0.05)")
```

```python

#Visualization
print(f"\nGenerating visualizations...")

fig = plt.figure(figsize=(16, 12))

# Plot 1: Complete time series with all satellites
ax1 = plt.subplot(3, 1, 1)
colors = {'Landsat 4': '#1f77b4', 'Landsat 5': '#ff7f0e', 
          'Landsat 7': '#2ca02c', 'Landsat 8': '#d62728', 
          'Landsat 9': '#9467bd'}

for satellite in df['satellite'].unique():
    sat_data = df[df['satellite'] == satellite]
    ax1.scatter(sat_data['date'], sat_data['mean_temp_C'], 
               label=satellite, alpha=0.6, s=15, color=colors.get(satellite, 'gray'))

ax1.set_xlabel('Date', fontsize=12, fontweight='bold')
ax1.set_ylabel('Surface Temperature (°C)', fontsize=12, fontweight='bold')
ax1.set_title('Lake Tahoe Surface Temperature Time Series (1982-2025)\nAll Landsat Missions', 
             fontsize=14, fontweight='bold')
ax1.legend(loc='best', framealpha=0.9)
ax1.grid(True, alpha=0.3, linestyle='--')

# Plot 2: Annual mean temperature with trend line
ax2 = plt.subplot(3, 1, 2)
ax2.plot(annual_means.index, annual_means.values, 'o-', 
        linewidth=2, markersize=8, alpha=0.7, label='Annual Mean')

# Add trend line
trend_line = slope * years + intercept
ax2.plot(years, trend_line, 'r--', linewidth=2, 
        label=f'Trend: {slope:.4f}°C/year (R²={r_value**2:.3f})')

ax2.set_xlabel('Year', fontsize=12, fontweight='bold')
ax2.set_ylabel('Mean Temperature (°C)', fontsize=12, fontweight='bold')
ax2.set_title('Annual Mean Surface Temperature with Linear Trend', 
             fontsize=14, fontweight='bold')
ax2.legend(loc='best', framealpha=0.9)
ax2.grid(True, alpha=0.3, linestyle='--')

# Plot 3: Seasonal patterns (monthly boxplot)
ax3 = plt.subplot(3, 1, 3)
df['month'] = df['date'].dt.month
monthly_data = [df[df['month'] == m]['mean_temp_C'].values for m in range(1, 13)]
bp = ax3.boxplot(monthly_data, labels=['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
                                        'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'],
                patch_artist=True)

for patch in bp['boxes']:
    patch.set_facecolor('lightblue')
    patch.set_alpha(0.7)

ax3.set_xlabel('Month', fontsize=12, fontweight='bold')
ax3.set_ylabel('Surface Temperature (°C)', fontsize=12, fontweight='bold')
ax3.set_title('Seasonal Temperature Patterns (All Years Combined)', 
             fontsize=14, fontweight='bold')
ax3.grid(True, alpha=0.3, linestyle='--', axis='y')

plt.tight_layout()
plt.savefig('lake_tahoe_temperature_analysis.png', dpi=300, bbox_inches='tight')
print(f" Plot saved as: lake_tahoe_temperature_analysis.png\n")



print("=" * 70)
print(f"\nOutput files created:")
print(f"  1. {csv_filename}")
print(f"  2. lake_tahoe_temperature_analysis.png")

plt.show()
```

Findings:
There is evidence of warming at Lake Tahoe in the graph.  Over the roughly 42-year period depicted (mid-1980s to late 2020s), the linear trend shows a temperature increase of 0.0585°C annually, or about 2.5°C of warming.  This is significant and consistent with more general patterns of global warming.  The upward-sloping red trendline makes it evident that recent years have consistently seen higher temperatures than previous decades, with the most recent measurements clustering around 12-14°C compared to 9-11°C in the late 1980s and 1990s. The blue data points show significant year-to-year variability.  The warming signal is statistically significant and alarming for the ecosystem of Lake Tahoe, even though the R2 value of 0.177 shows that only roughly 18% of the temperature variance can be explained by the linear trend alone.
The initial uptick in the graph was concerning for me. Upon research I found that since Lake Tahoe's water temperature would never actually fluctuate by 10°C in a single year, the first spike, which occurs around 1983–1984, when the temperature jumps from about 4.3°C to 14.8°C, is most likely a data artifact rather than a true climate signal.  During the early days of Landsat 4's Thematic Mapper sensor's operation, a number of factors likely contributed to this anomaly: early calibration issues, nascent atmospheric correction algorithms that failed to adequately account for water vapor and other atmospheric effects, possible cloud contamination in mixed pixels, and a lack of clear-sky observations that allowed inaccurate measurements to disproportionately affect the annual mean.  This is supported by the fact that temperatures soon return to more stable levels in the 9–13°C range.
